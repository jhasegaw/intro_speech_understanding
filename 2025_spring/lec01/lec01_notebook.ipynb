## 3. Installing github desktop

It is possible to interact with github using a command-line program called `git`.  However, authentication (proving to github that you are the valid owner of your account) is a little complicated using `git`.  The github desktop simplifies that somewhat.  Go to https://desktop.github.com/, and download the app.



Open the app, and choose `Sign in to Github.com`:



Choose `Authorize desktop`:



Enter your github account's password, and click `Confirm:`



Back in the app, you see the name of your account, and a weird long e-mail address.  This is an e-mail address that github will create for you.  If people see your repository online and want to contact you, they can send e-mail to this address, and github will forward the e-mail to you.  That way, you don't have to publicize your real e-mail address.  I recommend you keep these options, and just click `Finish:`



If you're on Macintosh, you might get this prompt window.  I recommend you click `Move and Restart`:



Now, finally, you have the startup window for the github desktop app:



Now we want to `clone` your repository.  That means to copy it from github to your laptop.  Click on the repository name, then click on `Clone <username>/intro_speech_understanding`.

Now you should see the following window.  

**Notice the local path** (in my case, it is `/Users/jhasegaw/Documents/Github/intro_speech_understanding`).  That's the location on your computer where all the files will be stored.  **You will need to know that path name!!**  Please write it down somewhere.

Then click `Clone`.



Under `How are you planning to use this repository`, choose

**For my own purposes**

**This is important!**  If you use the defaults, github will try to upload your code to the main course directory, which won't work.

Then click `Continue`.
